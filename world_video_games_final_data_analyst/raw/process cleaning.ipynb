# cleaning data world_video_games_final.csv

load file

```python
import pandas as pd
import numpy as np
import pyarrow
import fastparquet

df = pd.read_csv('world_video_games_final.csv')
print(df.info)
```

```python
df.info()
```

mengubah dtype release_date menjadi datetime48

```python
df['release_date'] = pd.to_datetime(df['release_date'])
```

```python
df.info()
```

mencari baris yang kosong

```python
df.isna().sum()
```

terdapat baris yang kosong pada kolom release_date namun karena release_date pada game tersebut belum diketahui jadi sebaiknya dibiarkan kosong

kemudian mengubah type baris pada kolom genre menjadi list agar dapat dianalisis setiap nama genrenya

```python
# 1. Pastikan bertipe teks dan potong berdasarkan koma
df['genres'] = df['genres'].astype(str).str.split(', ')

# 2. Pecah list menjadi baris baru
df_terpisah = df.explode('genres')

# 3. Rapikan nomor indeksnya
df_terpisah = df_terpisah.reset_index(drop=True)

# 4. Tampilkan hasilnya
print(df_terpisah.head())
```

kemudian mengubah type baris pada kolom platform menjadi list agar dapat dianalisis setiap nama platformnya

```python
# 1. Pastikan bertipe teks dan potong berdasarkan koma
df['platforms'] = df['platforms'].astype(str).str.split(', ')

# 2. Pecah list menjadi baris baru
df_split = df.explode('platforms')

# 3. Rapikan nomor indeksnya
df_split = df_split.reset_index(drop=True)

# 4. Tampilkan hasilnya
print(df_split.head())
```
